# preprocessing

```python
x = df.drop(columns=['customer_id',"loan_status"])
y = df["loan_status"]
```

```python
from sklearn.preprocessing import StandardScaler,RobustScaler,OneHotEncoder,LabelEncoder
from imblearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
```

```python
ro_col = ['savings_assets']
```

```python
st_cols = ['age', 'years_employed',
       'annual_income', 'credit_score', 'credit_history_years',
       'savings_assets', 'current_debt',
       'delinquencies_last_2yrs', 'derogatory_marks'
       , 'loan_amount', 'interest_rate', 'debt_to_income_ratio',
       'loan_to_income_ratio', 'payment_to_income_ratio']
```

```python
ohe_cols = ['occupation_status','product_type','loan_intent']
```

```python
# standard scaler 
st_pipe = Pipeline(steps=[
                    ("st_scaler",StandardScaler())
                    ])

st_pipe
```

```python
# Robust scaler
ro_pipe = Pipeline(steps=[
                    ("ro_scaler",RobustScaler()) 
                    ])

ro_pipe
```

```python
# ohe pipeline
ohe_pipe = Pipeline(steps=[
    ("ohe_encoder",OneHotEncoder(drop="first",sparse_output=False))
                    ])

ohe_pipe
```

```python
ready_col = ['defaults_on_file']
```

```python
preprocessing_pipeline = ColumnTransformer(transformers=[
    ("st_pipe",st_pipe,st_cols),
    ("ro_pipe",ro_pipe,ro_col),
    ("ohe_pipe",ohe_pipe,ohe_cols),
    ("ready_pipe", "passthrough", ready_col)
    ])

preprocessing_pipeline
```

# select best model

```python
import sys
!{sys.executable} -m pip install xgboost
```

```python
from sklearn.metrics import make_scorer, f1_score, recall_score, precision_score, accuracy_score
from sklearn.model_selection import cross_validate
from imblearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier

scoring = {
    'F1': make_scorer(f1_score),
    'Recall': make_scorer(recall_score),
    'Precision': make_scorer(precision_score),
    'Accuracy': make_scorer(accuracy_score)
}


models = [
    ('Logistic Regression', LogisticRegression(random_state=42, n_jobs=-1)),
    ('Knn', KNeighborsClassifier()),
    ('Gaussian Naive Bayes', GaussianNB()),
    ('Decision Tree', DecisionTreeClassifier(random_state=42)),
    ('Random Forest', RandomForestClassifier(random_state=42, n_jobs=-1)),
    ('Xgboost', XGBClassifier()),
]


for name, model in models:
    pipeline = Pipeline([
        ('Preprocessing', preprocessing_pipeline),
        ('Model', model)
    ])
    
    result  = cross_validate(
        pipeline,
        x, y,
        cv=5,
        scoring=scoring,
        return_train_score=True,
        n_jobs=-1
    )
    
    print(name)
    print('Train F1      :', result['train_F1'].mean().round(2)*100)
    print('Test F1       :', result['test_F1'].mean().round(2)*100)
    print('Train Recall  :', result['train_Recall'].mean().round(2)*100)
    print('Test Recall   :', result['test_Recall'].mean().round(2)*100)
    print('Train Precision:', result['train_Precision'].mean().round(2)*100)
    print('Test Precision :', result['test_Precision'].mean().round(2)*100)
    print('Train Accuracy :', result['train_Accuracy'].mean().round(2)*100)
    print('Test Accuracy  :', result['test_Accuracy'].mean().round(2)*100)
    print('Total Training Time :', result['fit_time'].sum().round(2))
    print('-'*50)
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import GridSearchCV, cross_validate
from sklearn.pipeline import Pipeline

# KNN model
knn_model = KNeighborsClassifier()

pipeline = Pipeline([
    ('Preprocessing', preprocessing_pipeline),  
    ('KNN', knn_model)
])


param_grid = {
    'KNN__n_neighbors': [3, 5, 7, 9],
    'KNN__weights': ['uniform', 'distance'],
    'KNN__p': [1, 2] 
}

grid = GridSearchCV(
    estimator=pipeline,
    param_grid=param_grid,
    scoring=scoring,
    refit='F1',
    cv=5,
    n_jobs=-1,
    verbose=1
)

grid.fit(x, y)

print("Best Params:", grid.best_params_)
print("Best CV score:", grid.best_score_)

best_model = grid.best_estimator_


cv_results = cross_validate(
    best_model,
    x,
    y,
    cv=5,
    scoring=scoring,
    return_train_score=True,
    n_jobs=-1
)

for metric in scoring.keys():
    print(f"Train {metric} :", cv_results[f'train_{metric}'].mean().round(3))
    print(f"Test {metric}  :", cv_results[f'test_{metric}'].mean().round(3))
    print('-'*30)
```
